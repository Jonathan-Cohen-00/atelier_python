```python
sales = [120, 95, 140, 180, 75, 160]

# 1. Number of orders, total and mean

print("Number of orders:", len(sales))
print("Total sales:", sum(sales))
print("Mean sales:", sum(sales) / len(sales))


# 2. Define classify_sales(value), then apply it to each amount

def classify_sales(value):
    if value >= 150:
        return "High"
    else :
        return "Standard"
       

print("Sales classifications:")
for amount in sales:
    print(f"{amount}: {classify_sales(amount)}")

# 3. Build the list of high amounts using a loop and a condition

high_amounts = []
for amount in sales:
    if classify_sales(amount) == "High":
        high_amounts.append(amount)

print("High amounts:", high_amounts)

# 4. Build and display report
```
